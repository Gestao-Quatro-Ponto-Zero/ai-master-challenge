```python
# Analisa as distribuições das indústrias e países dos clientes.

# Exibe as 15 indústrias com maior número de contas no dataframe 'accounts'.
print("Indústrias:")
print(accounts['industry'].value_counts().head(15))

# Exibe os 15 países com maior número de contas no dataframe 'accounts'.
print("\nPaíses:")
print(accounts['country'].value_counts().head(15))
```

```python
# Calcula o número médio de tickets de suporte por conta para churnados vs. retidos.
# Agrupa os tickets por 'account_id' e conta o número de tickets para cada conta.
tickets_por_conta = tickets.groupby('account_id').size().reset_index(name='num_tickets')

# Mescla este resumo com o dataframe 'accounts' para incluir a flag de churn.
accounts_tickets = accounts.merge(
    tickets_por_conta,
    on='account_id',
    how='left'
)

# Preenche valores NaN (que indicam contas sem tickets) com 0.
accounts_tickets['num_tickets'] = accounts_tickets['num_tickets'].fillna(0)

# Calcula a média do número de tickets para contas que churnaram e para as que foram retidas.
print("Média de tickets por conta (churnados vs. retidos):")
display(accounts_tickets.groupby('churn_flag')['num_tickets'].mean())
```

```python
# Calcula o número médio de escalonamentos por conta para churnados vs. retidos.
# Agrupa os tickets por 'account_id' e soma as flags de escalonamento ('escalation_flag') para cada conta.
escalations = tickets.groupby('account_id')['escalation_flag'].sum().reset_index()

# Mescla este resumo com o dataframe 'accounts' para incluir a flag de churn.
accounts_escalation = accounts.merge(
    escalations,
    on='account_id',
    how='left'
)

# Preenche valores NaN (que indicam contas sem escalonamentos) com 0.
accounts_escalation['escalation_flag'] = accounts_escalation['escalation_flag'].fillna(0)

# Calcula a média do número de escalonamentos para contas que churnaram e para as que foram retidas.
print("Média de escalonamentos por conta (churnados vs. retidos):")
display(accounts_escalation.groupby('churn_flag')['escalation_flag'].mean())
```

```python
# Calcula a pontuação média de satisfação por conta para churnados vs. retidos.
# Agrupa os tickets por 'account_id' e calcula a média das pontuações de satisfação ('satisfaction_score') para cada conta.
satisfaction = tickets.groupby('account_id')['satisfaction_score'].mean().reset_index()

# Mescla este resumo com o dataframe 'accounts' para incluir a flag de churn.
accounts_satisfaction = accounts.merge(
    satisfaction,
    on='account_id',
    how='left'
)

# Calcula a média das pontuações de satisfação para contas que churnaram e para as que foram retidas.
print("Média de satisfação por conta (churnados vs. retidos):")
display(accounts_satisfaction.groupby('churn_flag')['satisfaction_score'].mean())
```

```python
# Analisa o uso de features (contagem, duração, erros, beta) para contas churnadas vs. retidas.

# Agrega o uso por 'subscription_id' para resumir as métricas de uso de cada assinatura.
usage_per_subscription = usage.groupby('subscription_id').agg({
    'usage_count': 'sum',
    'usage_duration_secs': 'sum',
    'error_count': 'sum',
    'is_beta_feature': 'sum' # Soma para identificar se houve uso de feature beta
}).reset_index()

# Mescla o uso por assinatura com as informações da conta (account_id) para associar o uso a cada conta.
usage_accounts = subscriptions[['subscription_id', 'account_id']].merge(
    usage_per_subscription,
    on='subscription_id',
    how='left'
)

# Agrega o uso novamente, mas desta vez por 'account_id', para obter o uso total por conta.
usage_accounts = usage_accounts.groupby('account_id').agg({
    'usage_count': 'sum',
    'usage_duration_secs': 'sum',
    'error_count': 'sum',
    'is_beta_feature': 'sum' # Soma o uso de features beta por conta
}).reset_index()

# Mescla com a flag de churn ('churn_flag') das contas para diferenciar churnados de retidos.
usage_accounts = accounts[['account_id', 'churn_flag']].merge(
    usage_accounts,
    on='account_id',
    how='left'
)

# Preenche valores NaN com 0 para contas que não tiveram uso registrado.
usage_accounts = usage_accounts.fillna(0)

# Calcula a média das métricas de uso (contagem, duração, erros e uso de beta) para contas churnadas e retidas.
print("Média de uso de features (churnados vs. retidos):")
display(usage_accounts.groupby('churn_flag')[[
    'usage_count',
    'usage_duration_secs',
    'error_count',
    'is_beta_feature'
]].mean())
```

```python
# Compara MRR (Receita Recorrente Mensal), ARR (Receita Recorrente Anual), e flags de upgrade/downgrade para contas churnadas vs. retidas.
# Mescla informações de conta (account_id, churn_flag) com dados de assinatura (subscriptions).
accounts_subs = accounts[['account_id','churn_flag']].merge(
    subscriptions,
    on='account_id',
    how='left'
)

# Agrupa por 'churn_flag' e calcula a média das métricas financeiras e de upgrade/downgrade.
print("Média de MRR, ARR, e flags de upgrade/downgrade (churnados vs. retidos):")
display(accounts_subs.groupby('churn_flag_x')[[
    'mrr_amount',
    'arr_amount',
    'upgrade_flag',
    'downgrade_flag'
]].mean())
```

```python
# Calcula a taxa de churn média por frequência de cobrança para entender como o método de pagamento afeta o churn.
print("Taxa de churn por frequência de cobrança:")
display(subscriptions.groupby('billing_frequency')['churn_flag'].mean())
```

```python
# Calcula a taxa de churn média por nível de plano ('plan_tier') para identificar quais planos têm maior probabilidade de churn.
print("Taxa de churn por nível de plano:")
display(subscriptions.groupby('plan_tier')['churn_flag'].mean())
```

```python
# Analisa a relação entre o uso de features beta e a flag de churn para verificar se usuários beta churnam mais ou menos.

# Cria uma coluna 'beta_user' no dataframe `usage_accounts` que é True se a conta usou alguma feature beta (is_beta_feature > 0).
usage_accounts['beta_user'] = usage_accounts['is_beta_feature'] > 0

# Cria uma tabela cruzada para analisar a distribuição de churn de usuários beta vs. não beta.
beta_analysis = usage_accounts.groupby([
    'beta_user',
    'churn_flag'
]).size().unstack(fill_value=0)

print("Análise de churn para usuários beta vs. não beta:")
display(beta_analysis)
```

```python
# Descreve o uso de features beta para contas churnadas e retidas, mostrando estatísticas como média, desvio padrão, etc.
print("Estatísticas de uso de features beta (churnados vs. retidos):")
display(usage_accounts.groupby('churn_flag')['is_beta_feature'].describe())
```

```python
# Analisa o uso total de cada feature (por nome da feature) para contas churnadas vs. retidas.

# Seleciona 'account_id' a partir de 'subscription_id' para associar assinaturas a contas.
usage_subs = subscriptions[['subscription_id','account_id']]

# Mescla o dataframe de uso de features com as informações de conta.
usage_features = usage.merge(
    usage_subs,
    on='subscription_id',
    how='left'
)

# Mescla o resultado com a flag de churn das contas para diferenciar churnados de retidos.
usage_features = usage_features.merge(
    accounts[['account_id','churn_flag']],
    on='account_id',
    how='left'
)

# Agrupa por nome da feature e 'churn_flag' para somar o uso de cada feature e ver a distribuição entre churnados e retidos.
print("Uso de features por nome (churnados vs. retidos) - Top 20:")
display(usage_features.groupby([
    'feature_name',
    'churn_flag'
])['usage_count'].sum().unstack(fill_value=0).head(20))
```

```python
# Conta a frequência de cada feedback de churn, incluindo valores NaN (sem feedback), para entender as razões mais comuns.
print("Frequência dos feedbacks de churn:")
display(churn['feedback_text'].value_counts(dropna=False))
```

```python
# Mostra os feedbacks de churn por motivo de churn, criando uma categorização mais granular das razões de churn.
print("Feedbacks de churn por motivo de churn:")
display(churn.groupby('reason_code')['feedback_text'].value_counts())
```

```python
# Agrupa contas por buckets de uso e calcula a taxa de churn para cada bucket, identificando se contas com maior ou menor uso têm mais churn.

# Cria buckets de uso (Baixo, Médio, Alto, Muito Alto) com base nos quartis da contagem de uso.
usage_accounts['usage_bucket'] = pd.qcut(
    usage_accounts['usage_count'],
    q=4,
    labels=['Baixo', 'Médio', 'Alto', 'Muito Alto']
)

# Calcula a taxa de churn para cada bucket de uso.
print("Taxa de churn por bucket de uso:")
display(usage_accounts.groupby('usage_bucket')['churn_flag'].mean())
```

```python
# Agrupa contas por buckets de erro e calcula a taxa de churn para cada bucket, verificando se a ocorrência de erros afeta o churn.

# Cria buckets de erro (Baixo, Médio, Alto, Muito Alto) com base nos quartis da contagem de erros.
usage_accounts['error_bucket'] = pd.qcut(
    usage_accounts['error_count'],
    q=4,
    labels=['Baixo', 'Médio', 'Alto', 'Muito Alto']
)

# Calcula a taxa de churn para cada bucket de erro.
print("Taxa de churn por bucket de erro:")
display(usage_accounts.groupby('error_bucket')['churn_flag'].mean())
```

```python
# Analisa métricas de uso (contagem, duração, erros) com base no feedback de churn.

# Seleciona um resumo das métricas de uso por conta do dataframe `usage_accounts`.
usage_summary = usage_accounts[['account_id',
                                'usage_count',
                                'usage_duration_secs',
                                'error_count']]

# Mescla as informações de churn com o resumo de uso para associar o feedback de churn às métricas de uso.
churn_usage = churn.merge(
    usage_summary,
    on='account_id',
    how='left'
)

# Agrupa por feedback de churn e calcula a média das métricas de uso, ordenando pelo 'usage_count' de forma decrescente.
print("Média de métricas de uso por feedback de churn:")
display(churn_usage.groupby('feedback_text')[
    ['usage_count',
     'usage_duration_secs',
     'error_count']
].mean().sort_values(
    by='usage_count',
    ascending=False
))
```

```python
# Calcula a taxa de churn média por fonte de indicação, ordenando de forma decrescente para destacar as fontes com maior churn.
print("Taxa de churn por fonte de indicação:")
display(accounts.groupby('referral_source')['churn_flag'].mean().sort_values(ascending=False))
```

```python
# Calcula a taxa de churn média por indústria, ordenando de forma decrescente para identificar indústrias mais propensas ao churn.
print("Taxa de churn por indústria:")
display(accounts.groupby('industry')['churn_flag'].mean().sort_values(ascending=False))
```

```python
# Calcula a taxa de churn média por país, ordenando de forma decrescente para identificar países com maior churn.
print("Taxa de churn por país:")
display(accounts.groupby('country')['churn_flag'].mean().sort_values(ascending=False))
```

```python
# Calcula a taxa de churn média para contas em período de trial vs. não trial, para entender o impacto do trial no churn.
print("Taxa de churn (trial vs. não trial):")
display(accounts.groupby('is_trial')['churn_flag'].mean())
```

```python
# Contagem de contas churnadas/retidas por indústria, mostrando a distribuição do churn em diferentes setores.
print("Contagem de churn por indústria:")
display(accounts.groupby(['industry','churn_flag']).size().unstack(fill_value=0))
```

```python
# Contagem de contas churnadas/retidas por fonte de indicação, para entender quais canais trazem clientes mais propensos ao churn.
print("Contagem de churn por fonte de indicação:")
display(accounts.groupby(['referral_source','churn_flag']).size().unstack(fill_value=0))
```

```python
# Analisa o MRR (Receita Recorrente Mensal) e ARR (Receita Recorrente Anual) médios para contas churnadas vs. retidas.

# Seleciona 'account_id' e 'churn_flag' do dataframe 'accounts'.
churn_accounts = accounts[['account_id','churn_flag']]

# Calcula MRR e ARR médios por conta a partir das assinaturas, agrupando por 'account_id'.
account_revenue = subscriptions.groupby('account_id').agg({
    'mrr_amount':'mean',
    'arr_amount':'mean'
}).reset_index()

# Mescla as informações de churn com o resumo de receita por conta.
financial = churn_accounts.merge(
    account_revenue,
    on='account_id',
    how='left'
)

# Calcula a média de MRR e ARR para contas churnadas e retidas.
print("MRR e ARR médios (churnados vs. retidos):")
display(financial.groupby('churn_flag')[[
    'mrr_amount',
    'arr_amount'
]].mean())
```

```python
# Analisa a taxa de churn, a contagem de contas e o MRR médio por indústria.

# Mescla o dataframe de contas com as informações de receita por conta (MRR, ARR).
churn_by_industry = accounts.merge(
    account_revenue,
    on='account_id',
    how='left'
)

# Agrupa por indústria e calcula a taxa média de churn, a contagem de contas e o MRR médio, ordenando pela taxa de churn decrescente.
print("Análise de churn e receita por indústria:")
display(churn_by_industry.groupby('industry').agg({
    'churn_flag':'mean',
    'account_id':'count',
    'mrr_amount':'mean'
}).sort_values(
    by='churn_flag',
    ascending=False
))
```

```python
# Calcula o uso médio de features por indústria.

# Mescla o dataframe de contas com informações de indústria e a contagem total de uso por conta.
accounts_usage = accounts[['account_id','industry']].merge(
    usage_accounts[['account_id','usage_count']],
    on='account_id',
    how='left'
)

# Agrupa por indústria e calcula o uso médio de features, ordenando de forma decrescente.
print("Uso médio de features por indústria:")
display(accounts_usage.groupby('industry')['usage_count'].mean().sort_values(
    ascending=False
))
```

### Cruzar Canal de Indicação com Motivo de Churn

Para entender a relação entre o canal de indicação (`referral_source`) e os motivos de churn (`reason_code`), vamos primeiro mesclar os dataframes `accounts` e `churn`.

```python
# Mescla os dataframes 'accounts' e 'churn' para incluir apenas contas que realmente churnaram, utilizando um 'inner merge'.
accounts_churn_merged = accounts.merge(
    churn,
    on='account_id',
    how='inner' # Apenas contas que realmente churnaram
)
print("Primeiras linhas do dataframe mesclado 'accounts_churn_merged':")
display(accounts_churn_merged.head())
```

Agora, vamos agrupar os dados por `referral_source` e `reason_code` para contar quantas vezes cada combinação ocorre. Isso nos permitirá identificar o principal motivo de churn para cada canal de indicação.

```python
# Agrupa os dados por 'referral_source' e 'reason_code' e conta a ocorrência de cada combinação.
# Isso nos permitirá identificar o principal motivo de churn para cada canal de indicação.
churn_reason_by_referral = accounts_churn_merged.groupby(['referral_source', 'reason_code']).size().unstack(fill_value=0)
print("Ocorrência de motivos de churn por canal de indicação:")
display(churn_reason_by_referral)
```

Finalmente, vamos identificar o principal motivo de churn para cada `referral_source` (canal de indicação) com base nos counts que obtivemos.

```python
# Identifica o motivo de churn mais frequente para cada 'referral_source' (canal de indicação) com base nos counts obtidos.
main_churn_reason_by_referral = churn_reason_by_referral.idxmax(axis=1)
print("Principal motivo de churn por canal de indicação:")
display(main_churn_reason_by_referral)
```

```python
# Analisa a proporção de uso de cada feature por contas que churnaram, para identificar features críticas.

# Prepara dados de uso, mesclando 'subscriptions' para obter 'account_id' e 'accounts' para 'churn_flag'.
usage_subs = subscriptions[['subscription_id', 'account_id']]
usage_features = usage.merge(usage_subs, on='subscription_id', how='left')

usage_features = usage_features.merge(
    accounts[['account_id', 'churn_flag']],
    on='account_id',
    how='left'
)

# Agrupa por feature e calcula o uso total e o uso por contas que churnaram.
feature_churn = usage_features.groupby('feature_name').agg(
    total_usage=('usage_count', 'sum'),
    churned_usage=('usage_count', lambda x: x[usage_features.loc[x.index, 'churn_flag'] == True].sum())
)

# Calcula a parcela do uso total que vem de contas churnadas.
feature_churn['churned_usage_share'] = feature_churn['churned_usage'] / feature_churn['total_usage']

# Exibe as 10 principais features com maior participação de uso por churnados, ordenadas de forma decrescente.
print("Top 10 features com maior participação de uso por contas churnadas:")
display(feature_churn.sort_values('churned_usage_share', ascending=False).head(10))
```

```python
# Descreve as estatísticas de uso (usage_count) para contas churnadas vs. retidas, fornecendo um resumo estatístico.
print("Estatísticas de 'usage_count' para contas churnadas vs. retidas:")
display(usage_accounts.groupby('churn_flag')['usage_count'].describe())
```

```python
# Calcula o uso médio por indústria e status de churn.

# Mescla o dataframe de contas com informações de indústria, 'churn_flag' e 'usage_count' por conta.
accounts_usage = accounts[['account_id','industry','churn_flag']].merge(
    usage_accounts[['account_id','usage_count']],
    on='account_id',
    how='left'
)

# Agrupa por indústria e 'churn_flag', calculando o uso médio de features para cada grupo.
print("Uso médio por indústria e status de churn:")
display(accounts_usage.groupby(['industry','churn_flag'])['usage_count'].mean())
```

```python
# Analisa a contagem de churns e o MRR total por indústria.

# Mescla o dataframe de contas com as informações de receita por conta.
industry_revenue = accounts.merge(
    account_revenue,
    on='account_id',
    how='left'
)

# Agrupa por indústria e calcula o total de churns (soma de 'churn_flag' onde True=1) e o MRR total.
print("Contagem de churns e MRR total por indústria:")
display(industry_revenue.groupby('industry').agg({
    'churn_flag':'sum',
    'mrr_amount':'sum'
}))
```

```python
# Calcula a taxa de churn (contagem total, soma de churns e taxa média) por indústria, ordenando pela taxa média de churn.
print("Taxa de churn por indústria (contagem, soma, média):")
display(accounts.groupby('industry')['churn_flag'].agg([
    'count',
    'sum',
    'mean'
]).sort_values('mean', ascending=False))
```

```python
# Calcula a taxa de churn (contagem total, soma de churns e taxa média) por fonte de indicação, ordenando pela taxa média de churn.
print("Taxa de churn por fonte de indicação (contagem, soma, média):")
display(accounts.groupby('referral_source')['churn_flag'].agg([
    'count',
    'sum',
    'mean'
]).sort_values('mean', ascending=False))
```

```python
# Analisa a contagem de churns por mês.

# Converte as colunas de data para o formato datetime para permitir operações de tempo.
churn['churn_date'] = pd.to_datetime(churn['churn_date'])
accounts['signup_date'] = pd.to_datetime(accounts['signup_date'])

# Agrupa os eventos de churn por mês e conta o número de churns em cada mês.
churn_by_month = churn.groupby(churn['churn_date'].dt.to_period('M')).size()

print("Contagem de churns por mês:")
display(churn_by_month)
```

```python
# Calcula a taxa de churn por coorte de mês de signup, para entender o comportamento de churn ao longo do tempo para diferentes grupos de clientes.

# Extrai o mês de signup para cada conta e cria uma nova coluna 'signup_month'.
accounts['signup_month'] = accounts['signup_date'].dt.to_period('M')

# Agrupa por mês de signup e calcula a contagem total de contas e o número de contas churnadas em cada coorte.
cohort_churn = accounts.groupby('signup_month').agg(
    accounts=('account_id', 'count'),
    churned=('churn_flag', 'sum')
)

# Calcula a taxa de churn para cada coorte (contas churnadas / total de contas).
cohort_churn['churn_rate'] = cohort_churn['churned'] / cohort_churn['accounts']

# Exibe as coortes ordenadas por mês de signup para visualizar a tendência.
print("Taxa de churn por coorte de mês de signup:")
display(cohort_churn.sort_values('signup_month'))
```

```python
# Analisa o uso de features nos 90 dias antes do churn para identificar padrões de comportamento de uso que antecedem o churn.

# Converte a coluna 'usage_date' para o formato datetime.
usage['usage_date'] = pd.to_datetime(usage['usage_date'])

# Mescla dados de uso com informações de conta (via 'subscriptions') e churn.
usage_with_account = usage.merge(
    subscriptions[['subscription_id', 'account_id']],
    on='subscription_id',
    how='left'
)

churn_usage_timeline = usage_with_account.merge(
    churn[['account_id', 'churn_date']],
    on='account_id',
    how='inner'
)

# Calcula o número de dias antes do churn para cada registro de uso.
churn_usage_timeline['days_before_churn'] = (
    churn_usage_timeline['churn_date'] - churn_usage_timeline['usage_date']
).dt.days

# Filtra o uso que ocorreu nos 90 dias antes do churn (dias_before_churn entre 0 e 90).
pre_churn_usage = churn_usage_timeline[
    (churn_usage_timeline['days_before_churn'] >= 0) &
    (churn_usage_timeline['days_before_churn'] <= 90)
]

# Soma o 'usage_count' em buckets de 30 dias antes do churn para uma análise agregada.
print("Soma de 'usage_count' em buckets de 30 dias antes do churn:")
display(pre_churn_usage.groupby(
    pd.cut(pre_churn_usage['days_before_churn'], bins=[0, 30, 60, 90], right=False, labels=['0-29 dias', '30-59 dias', '60-89 dias'])
)['usage_count'].sum())
```

```python
# Analisa a amplitude de uso de features (número de features únicas usadas) para contas churnadas vs. retidas.

# Calcula o número de features únicas usadas por cada conta.
feature_breadth = usage_with_account.groupby('account_id')['feature_name'].nunique().reset_index()
feature_breadth.columns = ['account_id', 'unique_features_used']

# Mescla este resumo com o dataframe 'accounts' para incluir a flag de churn.
feature_breadth_accounts = accounts[['account_id', 'churn_flag']].merge(
    feature_breadth,
    on='account_id',
    how='left'
)

# Preenche contas sem uso de features (que não apareceram em `usage_with_account`) com 0.
feature_breadth_accounts['unique_features_used'] = feature_breadth_accounts['unique_features_used'].fillna(0)

# Descreve a amplitude de uso de features para contas churnadas e retidas, fornecendo estatísticas comparativas.
print("Estatísticas da amplitude de uso de features (churnados vs. retidos):")
display(feature_breadth_accounts.groupby('churn_flag')['unique_features_used'].describe())
```

```python
# Resumo das métricas de suporte para contas churnadas vs. retidas.

# Converte a coluna 'submitted_at' para o formato datetime.
tickets['submitted_at'] = pd.to_datetime(tickets['submitted_at'])

# Agrega as métricas de suporte por conta, calculando o número de tickets, tempo médio de resolução, tempo médio da primeira resposta, número de escalonamentos e satisfação média.
support_summary = tickets.groupby('account_id').agg(
    tickets=('ticket_id', 'count'),
    avg_resolution_hours=('resolution_time_hours', 'mean'),
    avg_first_response_minutes=('first_response_time_minutes', 'mean'),
    escalations=('escalation_flag', 'sum'),
    avg_satisfaction=('satisfaction_score', 'mean')
).reset_index()

# Mescla este resumo com o dataframe 'accounts' para incluir a flag de churn.
support_accounts = accounts[['account_id', 'churn_flag']].merge(
    support_summary,
    on='account_id',
    how='left'
)

# Calcula a média das métricas de suporte para contas churnadas e retidas para comparação.
print("Média das métricas de suporte (churnados vs. retidos):")
display(support_accounts.groupby('churn_flag')[
    ['tickets', 'avg_resolution_hours', 'avg_first_response_minutes', 'escalations', 'avg_satisfaction']
].mean())
```

```python
# Prepara o dataframe para modelagem (englobando várias métricas e one-hot encoding).

# Mescla o dataframe 'accounts' com a amplitude de features ('feature_breadth'), o resumo de suporte ('support_summary') e a receita por conta ('account_revenue').
model_df = accounts.merge(feature_breadth, on='account_id', how='left')
model_df = model_df.merge(support_summary, on='account_id', how='left')
model_df = model_df.merge(account_revenue, on='account_id', how='left')

# Preenche valores ausentes (NaN) com 0, assumindo que a ausência de dados significa zero ocorrência ou valor.
model_df = model_df.fillna(0)

# Aplica One-Hot Encoding para variáveis categóricas ('industry', 'referral_source', 'plan_tier', 'country'), criando novas colunas binárias e removendo a coluna original para evitar multicolinearidade.
model_df = pd.get_dummies(
    model_df,
    columns=['industry', 'referral_source', 'plan_tier', 'country'],
    drop_first=True
)

print("Primeiras linhas do dataframe preparado para modelagem:")
display(model_df.head())
```

```python
# Calcula a taxa de churn mensal ao longo do tempo.

# Converte as colunas 'start_date' e 'end_date' em 'subscriptions' para o formato datetime.
subscriptions['start_date'] = pd.to_datetime(subscriptions['start_date'])
subscriptions['end_date'] = pd.to_datetime(subscriptions['end_date'])

# Define o intervalo de meses a serem analisados, do início da primeira assinatura até o último evento de churn.
months = pd.period_range(
    subscriptions['start_date'].min().to_period('M'),
    churn['churn_date'].max().to_period('M'),
    freq='M'
)

monthly_active = []

# Itera por cada mês para calcular o número de contas ativas e churnadas.
for month in months:
    month_start = month.to_timestamp()
    month_end = month.to_timestamp(how='end')

    # Contas ativas no mês: aquelas que começaram antes ou no final do mês E (não terminaram OU terminaram depois ou no início do mês).
    active = subscriptions[
        (subscriptions['start_date'] <= month_end) &
        ((subscriptions['end_date'].isna()) | (subscriptions['end_date'] >= month_start))
    ]['account_id'].nunique()

    # Contas churnadas no mês: aquelas cujo evento de churn ocorreu neste mês.
    churned = churn[churn['churn_date'].dt.to_period('M') == month]['account_id'].nunique()

    monthly_active.append({
        'month': month,
        'active_accounts': active,
        'churned_accounts': churned,
        'churn_rate': churned / active if active > 0 else 0 # Calcula a taxa de churn.
    })

# Cria um dataframe com as taxas de churn mensais e exibe as últimas 12 linhas.
monthly_churn_rate = pd.DataFrame(monthly_active)
print("Últimas 12 linhas da taxa de churn mensal ao longo do tempo:")
display(monthly_churn_rate.tail(12))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Define as features (variáveis independentes) para o modelo de machine learning, excluindo IDs e datas.
features = [
    col for col in model_df.columns
    if col not in ['account_id','account_name','signup_date','signup_month','churn_flag']
]

X = model_df[features] # Variáveis independentes para o modelo.
y = model_df['churn_flag'].astype(int) # Variável dependente (target), convertida para inteiro (0 ou 1).

# Divide os dados em conjuntos de treino (75%) e teste (25%), mantendo a proporção da variável 'churn_flag' em ambos os conjuntos (stratify=y).
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Cria um pipeline para o modelo: primeiro, escala os dados usando StandardScaler, e depois aplica a Regressão Logística.
model = Pipeline([
    ('scaler', StandardScaler()),
    ('logreg', LogisticRegression(max_iter=1000)) # Define o número máximo de iterações para convergência.
])

# Treina o modelo usando os dados de treino.
model.fit(X_train, y_train)

# Calcula as probabilidades de previsão de churn no conjunto de teste. O `[:, 1]` seleciona a probabilidade da classe positiva (churn).
pred_proba = model.predict_proba(X_test)[:, 1]

# Avalia o modelo usando a métrica ROC-AUC (Receiver Operating Characteristic - Area Under the Curve).
print("ROC-AUC:", roc_auc_score(y_test, pred_proba))

# Exibe um relatório de classificação detalhado, incluindo precisão, recall, f1-score e suporte.
print("\nRelatório de Classificação:")
print(classification_report(y_test, model.predict(X_test)))
```

```python
# Exibe os coeficientes do modelo de regressão logística para entender a importância e a direção de impacto de cada feature.

# Cria um dataframe com as features e seus respectivos coeficientes, ordenando pelos coeficientes de forma decrescente.
coefs = pd.DataFrame({
    'feature': features,
    'coefficient': model.named_steps['logreg'].coef_[0]
}).sort_values('coefficient', ascending=False)

print("Top 15 features por coeficiente do modelo de regressão logística:")
display(coefs.head(15))
```
